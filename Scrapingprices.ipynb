Our objective here is to use an llm to scrape prices from amazon and flipkart.Typically this is done through beautiful soup in python, [here](https://www.youtube.com/watch?v=qukjS96clB8&pp=ygUkd2ViIHNjcmFwaW5nIGVjb21tZXJjZSBwcmljZSB0cmFja2Vy) is a tutorial explaining the same.


```python
#lets try to get prices of tvs keyboards from amazon
import requests
url = 'https://web.archive.org/web/20070601180948/http://amazon.com/gp/bestsellers/books/'
headers2 = {'User-Agent': 'Mozilla/5.0 (X11; Linux x86_64; rv:132.0) Gecko/20100101 Firefox/132.0','Accept-Language':'en-US,en;q=0.5'}
response = requests.get(url, headers=headers2)
```

```python
len(response.content.decode('utf-8',errors='ignore'))#extremely large web pages, looks like this is very common phenomenon.
```

```python
import torch
from transformers import pipeline

model_id = "Qwen/Qwen2-7B-Instruct"
pipe = pipeline(
    "text-generation",
    model=model_id,
    torch_dtype=torch.bfloat16,
    device_map="cuda",
)
messages = [
    {"role": "system", "content": f"You are a concise webscraping assistant"},
    {"role": "user", "content": f"it is a amazon web page on books,dig into the webpage and give me the names of books and their corresponding prices in a csv format with columns [NAME, AUTHOR, PRICE] format them into a pretty table like format, Page: {response.content.decode('utf-8',errors='ignore')}"},
    ]
outputs = pipe(
    messages,
    max_new_tokens=2048,
)
print(outputs[0]["generated_text"][-1]['content'])
```

Although this works,it took lot of workarounds to acheve this and it is very hard with llms as even small webpages exceed the context length of the model (typically is around 128k)
So this is not practical at the moment.
